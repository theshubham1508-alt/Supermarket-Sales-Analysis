```python
import pandas as pd
df = pd.read_excel("SUPER MARKET DATA.xlsx")
df.head()
```

```python
print("Rows and Columns:", df.shape)
df
```

```python
print("\nColumn Names:")
df
```

```python
print(df.columns.tolist())
df
```

```python
print("\nMissing Values:")
df
print(df.isnull().sum())
df
```

```python
print("\nDuplicate Rows:", df.duplicated().sum())
df
```

```python
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)
df
```

```python
print("Highest Selling Product:")
print(product_sales.head(10))
```

```python
branch_sales = df.groupby("Branch")["Sales"].sum().sort_values(ascending=False)
```

```python
print("\nBranch-wise Sales:")
print(branch_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
print("\nCategory-wise Sales:")
print(category_sales)
```

```python
payment_count = df["Payment"].value_counts()

print("\nPayment Method Usage:")
print(payment_count)
```

```python
customer_sales = df.groupby("Customer Type")["Sales"].mean()

print("\nAverage Spending:")
print(customer_sales)
```

```python
average_rating = df["Rating"].mean()

print("\nAverage Customer Rating:", round(average_rating, 2))
```

```python
import matplotlib.pyplot as plt
product_sales = df.groupby("Product")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
product_sales.plot(kind="bar")
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
branch_sales = df.groupby("Branch")["Sales"].sum()
plt.figure(figsize=(7,5))
branch_sales.plot(kind="bar")
plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(9,5))
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_count = df["Payment"].value_counts()
plt.figure(figsize=(7,5))
payment_count.plot(kind="bar")
plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
payment_count = df["Payment"].value_counts()
plt.figure(figsize=(7,5))
payment_count.plot(kind="bar")
plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```
